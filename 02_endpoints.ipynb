```python
# ===== Notebook 02: Dual-Endpoint Simulation =====
from src.randomization import stratified_block_randomization
from src.simulate import simulate_baseline, merge_with_randomization, simulate_endpoints

# Rebuild data up to DATA2
random_df = stratified_block_randomization(seed=1128)
data1 = simulate_baseline(random_df, seed=1128)
data2 = merge_with_randomization(data1, random_df)

# Simulate endpoints
data3 = simulate_endpoints(data2, seed=1128)
print("DATA3 shape:", data3.shape)

# Verify correlations
r_ep = data3[['FEV1_Trough_D15', 'FEV1_AUC_1_4h_D15']].corr().iloc[0, 1]
r_cov = data3['FEV1_Trough_D15'].corr(data3['FEV1_FVC'])
print(f"\nEndpoint-endpoint correlation: {r_ep:.3f} (target 0.6)")
print(f"Endpoint vs baseline FEV1/FVC: {r_cov:.3f} (target ~0.3)")

# Show first rows
display(data3[['ID', 'Group', 'FEV1_Trough_D15', 'FEV1_AUC_1_4h_D15']].head())

# Save
data3.to_csv('outputs/tables/DATA3_endpoints.csv', index=False)
print("\nDATA3 saved.")
```
